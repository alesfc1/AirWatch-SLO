```python
import pandas as pd
import glob
import re
import os
from io import StringIO
```

```python
data_path = "/content/drive/MyDrive/MOS projekt 2025/DATASETS/"
```

```python
# iteracija skozi vse dokumente .csv
all_files = glob.glob(os.path.join(data_path, "*.csv"))
```

```python
def process_file(file_path):
    print(f"Obdelava {os.path.basename(file_path)} ...")

    # pridobivanje imen občin iz imena datoteke
    filename = os.path.basename(file_path)
    names = re.findall(r"'([^']+)'", filename)

    with open(file_path, "r", encoding="utf-8") as f:
        lines = f.readlines()

    data_start_idx = None
    for i, line in enumerate(lines):
        if line.strip().startswith("location_id,time"):
            data_start_idx = i
            break

    if data_start_idx is None:
        raise ValueError(f"Glave podatkov ni bilo mogoče najti v {file_path}")

    # koordinate
    meta_lines = [line for line in lines[:data_start_idx] if line.strip()]
    meta_part = "".join(meta_lines)
    meta_df = pd.read_csv(StringIO(meta_part))

    data_part = "".join(lines[data_start_idx:]).strip()
    df = pd.read_csv(StringIO(data_part))

    if len(names) != meta_df.shape[0]:
        print(f"⚠️ Opozorilo: neskladje med številom imen in koordinatami v {filename}")
    meta_df['municipality'] = names[:len(meta_df)]

    df = df.merge(meta_df[['location_id', 'latitude', 'longitude', 'municipality']], on='location_id', how='left')

    df.columns = [c.strip().lower().replace(" ", "_") for c in df.columns]
    df['time'] = pd.to_datetime(df['time'], errors='coerce')
    df['year'] = df['time'].dt.year

    pollutants = ['pm10_(μg/m³)', 'pm2_5_(μg/m³)', 'ozone_(μg/m³)',
                  'sulphur_dioxide_(μg/m³)', 'nitrogen_dioxide_(μg/m³)',
                  'carbon_monoxide_(μg/m³)']
    available_pollutants = [p for p in pollutants if p in df.columns]

    annual = (
        df.groupby(['municipality', 'year'])[available_pollutants + ['latitude', 'longitude']]
        .mean()
        .reset_index()
    )

    annual = annual.round(3)

    rename_map = {
        'municipality': 'Občina',
        'year': 'Leto',
        'latitude': 'Zemljepisna širina',
        'longitude': 'Zemljepisna dolžina',
        'pm10_(μg/m³)': 'PM10 (µg/m³)',
        'pm2_5_(μg/m³)': 'PM2.5 (µg/m³)',
        'ozone_(μg/m³)': 'Ozon (µg/m³)',
        'sulphur_dioxide_(μg/m³)': 'Žveplov dioksid (µg/m³)',
        'nitrogen_dioxide_(μg/m³)': 'Dušikov dioksid (µg/m³)',
        'carbon_monoxide_(μg/m³)': 'Ogljikov monoksid (µg/m³)',
    }

    annual = annual.rename(columns=rename_map)

    return annual
```

```python
all_results = []
for f in all_files:
    processed = process_file(f)
    if not processed.empty:
        all_results.append(processed)
```

```python
final_df = pd.concat(all_results, ignore_index=True)

final_df = final_df.drop(['Zemljepisna širina', 'Zemljepisna dolžina'], axis=1)
```

```python
output_path = ("podatki_kakovost_zraka.csv")
final_df.to_csv(output_path, index=False)

print("Letni podatki so bili uspešno združeni in shranjeni v:", output_path)
```
